```python
import pandas as pd
import numpy as np
```

```python
data={
    'name':['sitaa','tarun','anu','ashu','vinay','vishwa','parvati','kshama','veena','ashwini','harsha','akshara','ananth','nischal','akash'],
    'age':[18,19,np.nan,21,17,np.nan,18,18,19,np.nan,21,18,19,21,21],
    'marks':[99,100,55,78,98,65,69,78,89,58,33,26,89,90,56],
    'city':['mysore','ramangar','hubli','haveri',np.nan,'bnglr','madya','coorg','hubli','mysore',np.nan,'ooty','davanagere','goa','hassan'],

}
df=pd.DataFrame(data)
print("dataset:")
print(df)
```

```python
print('\nShape of dataset (rows,columns):',df.shape)
```

```python
print('\nColumn names:',df.columns.tolist())
```

```python
print('\nData types of each column:')
df.info()
```

```python
print("\nDescriptive statistics (numerical columns):")
print(df.describe())
```

```python
print('\nmean:\n',df.mean(numeric_only=True))
```

```python
print('\nMedian:\n',df.median(numeric_only=True))
```

```python
print('\nStandard deviation:\n',df.std(numeric_only=True))
```

```python
print('\nmissing values in each columns:')
print(df.isnull().sum())
print('\nTotal missing values in dataset:',df.isnull().sum().sum())
```

```python
print('\nDuplicate rows in dataset:')
print(df[df.duplicated()])
print('\nNumber of duplicate rows:',df.duplicated().sum())
print('\nEDA process completed successfully!')
```
